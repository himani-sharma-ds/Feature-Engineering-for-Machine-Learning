# Lable Encoder 

```python
import pandas as pd
```

```python
df = pd.DataFrame({"Name": ["Aman","Nikita","Ajay","OM","Priya","Arpit","Himani","Aman","Nikita","Ajay","OM","Priya","Arpit","Himani"]})
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
le.fit_transform(df["Name"])
```

```python
df["en_name"] = le.fit_transform(df["Name"])
```

```python
df
```

```python
new_data = df
new_data.to_csv("Label_Encoder.csv",index=False)
```
